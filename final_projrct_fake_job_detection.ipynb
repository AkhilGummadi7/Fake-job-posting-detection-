## **Importing Libraries**

```python
# ==============================
# Fake Job Posting Detection
# Part 1 - Import Libraries
# ==============================
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")

pd.set_option("display.max_columns",None)
sns.set_style("whitegrid")
```

## **Load The Dataset**

```python
df = pd.read_csv("/content/Fake_Job_Posting_Detection_Dataset.csv", engine='python')

print(df.shape)

df.head()
```

## **Basic info**

```python
df.info()
```

```python
df.describe()
```

## **Check Missing Values**

```python
missing = pd.DataFrame(df.isnull().sum(),columns=["Missing Values"])

missing["Percentage"] = (missing["Missing Values"]/len(df))*100

missing.sort_values("Percentage",ascending=False)
```

## **Visualize Missing Values**

```python
plt.figure(figsize=(12,6))

sns.boxplot(df.isnull().sum(axis=1))

plt.title("Missing Value Heatmap")

plt.show()
```
